Le problème consiste à simuler une marche aléatoire en 1D avec une
probabilité p que le pas soit +1 et une probabilité 1 - p qu’il soit -1.

On va tester plusieurs méthodes de NumPy pour montrer que cela facilite
beaucoup les calculs et rend l’exécution plus rapide.


```python
import numpy
```

```python
#1 

def marche(p, N):

    #génére N nombres aléatoires entre 0 et 1, Pour chaque nombre, choisit +1 si < p, sinon -1 (prob. p et q)

    steps = numpy.where(numpy.random.random(N) < p, 1, -1)
    deplacement =  numpy.sum(steps) #calcule la somme totale des steps (deplacement)
    # print(steps)

    return deplacement

#Valeurs pour le test

n1 = 200
p1 = 0.1
n2 = 100
p2 = 0.5
n3 = 45
p3 = 0.9
n4 = 10000
p4 = 0.2

d1 = marche(p1, n1)
d2 = marche(p2, n2)
d3 = marche(p3, n3)
d4 = marche(p4, n4)

print('Le déplacement total pour chaque essai a été de', d1, d2, d3, d4, 'unités.')
print('déplacement de' ,d1, 'unités aprés', n1, 'steps' )
print('déplacement de' ,d2, 'unités aprés', n2, 'steps' )
print('déplacement de' ,d3, 'unités aprés', n3, 'steps' )
print('déplacement de' ,d4, 'unités aprés', n4, 'steps' )
```

```python
#2

def monte_carlo(p, N, M):

    deplacements = numpy.zeros(M) #Crée un tableau de zéros pour stocker les déplacements finaux
    for i in range(M):
        
        deplacements[i] = marche(p, N) #remplit le tableau avec les résultats de chaque marche

    #fonctions statistiques
    moyenne = numpy.mean(deplacements)
    ecart_type = numpy.std(deplacements)
    numpy.save('deplacements_pedro.npy', deplacements)

    return moyenne, ecart_type

m = 900
n = 50
p = 0.9

moy, ecart_t = monte_carlo(p,n,m)

print('moyenne pour p =',p,'et N =',n, ':', moy)
print('ecart type pour p =',p,'et N =',n, ':', ecart_t)
```

```python
#3

def variation_M(p, N, M_values):

    resultats = numpy.zeros((2, 4)) #crée le tableau 2x4 pour stocker les résultats
    
    #remplit chaque colonne avec moyenne (ligne 0) et écart-type (ligne 1) pour chaque M
    
    resultats[0, 0], resultats[1, 0] = monte_carlo(p, N, M_values[0])
    resultats[0, 1], resultats[1, 1] = monte_carlo(p, N, M_values[1])
    resultats[0, 2], resultats[1, 2] = monte_carlo(p, N, M_values[2])
    resultats[0, 3], resultats[1, 3] = monte_carlo(p, N, M_values[3])
    
    numpy.save('resultats_marche_pedro.npy', resultats) #sauvegarde les résultats
    return resultats

p = 0.5
n = 1000
M_test = [10, 100, 1000, 10000]

resultats = variation_M(p, n, M_test)

print(resultats)
```
