Le problème consiste à simuler une marche aléatoire en 1D avec une
probabilité p que le pas soit +1 et une probabilité 1 - p qu’il soit -1.

On va tester plusieurs méthodes de NumPy pour montrer que cela facilite
beaucoup les calculs et rend l’exécution plus rapide.


```python
import numpy
```

```python
#1 - Sans bouclé

def marche_aleatoire(N,p):

    #le prochaine ligne crée un tableau de N step, elle met 1 si un nombre aléatoire est plus petit que p, et -1 sinon(probabilités p et q).
    ale_walk = numpy.where(numpy.random.random(N) < p, 1, -1)
    # print(ale_walk)
    deplacement = numpy.sum(ale_walk) #cette part on utilise le function numpy.sum pour calculer le deplacement total
    return deplacement


p1 = 0.5
n1 = 10

print(marche_aleatoire(n1,p1))
```

```python
#test avec plusieurs valeurs

#x = [N,p]
x = [[10,0.9],[50,0.3],[12,0.5],[1500,0.1],[200,0.99]]

for k in range((len(x))):

    dep = marche_aleatoire(x[k][0],x[k][1])
    print(f'Le déplacement total de {dep} unités pour N = {x[k][0]} et p = {x[k][1]}')
```

```python
#2

def monte_carlo_method(p, N, M):
   
    #tableau pour stocker les M résultats
    deplacements_finaux = numpy.zeros(M)
    
    #M fois
    for i in range(M):

        # stocke le déplacement final de chaque loop
        deplacements_finaux[i] = marche_aleatoire(N, p)
        
    # fonctions statistiques
    moyenne = numpy.mean(deplacements_finaux)
    ecart_t = numpy.std(deplacements_finaux)
    
    return moyenne, ecart_t
```

```python
#test

p = 0.99
N = 100
M = 1000  

print(f"Parametres : p={p}, N={N}, M={M}")

moyenne_mc, ecart_t_mc = monte_carlo_method(p, N, M)

print(f"\nRésultats de la simulation :")
print(f"Moyenne : {moyenne_mc:.4f}")
print(f"Écart: {ecart_t_mc:.4f}")
```

```python
3#


M = [10, 100, 1000, 10000]
nb_colonnes = len(M)

# 2. Criar o 'tableau' de resultados (2 linhas, 4 colunas)
# numpy.zeros
resultats = numpy.zeros((2, nb_colonnes))

# 3. Fazer a simulação para cada valor de M
for i, M_val in enumerate(M):
    
    # Chama a função que usa numpy.mean e numpy.std
    moyenne, ecart_type = monte_carlo_method(p, N, M_val)
    
    # Preenche o 'tableau'
    resultats[0, i] = moyenne  # Linha 0 : médias
    resultats[1, i] = ecart_type # Linha 1 : desvios-padrão

# 4. Mostrar e salvar o 'tableau'
print("\n--- 'Tableau' de resultados (Linha 0 = Médias, Linha 1 = Desvios-padrão) ---")
print(resultats)

# 5. Salvar o 'tableau' em um arquivo .npy
# nom_fichier = 'resultat_marche_aleatoire.npy'
# numpy.save
# numpy.save(nom_fichier, resultats)

# print(f"\n'Tableau' salvo com sucesso no arquivo: {nom_fichier}")
```
